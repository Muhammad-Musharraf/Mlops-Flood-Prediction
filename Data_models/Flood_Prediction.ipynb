```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
```

```python
x_train_scaled=scaler.transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor

```

```python
lr=LinearRegression()
knn=KNeighborsRegressor()
Lr=LogisticRegression()
dt=DecisionTreeRegressor()
rf=RandomForestRegressor()
svc=SVR()
rd=Ridge(alpha=1.0)
las=Lasso()
gb=GradientBoostingRegressor()


xgb=XGBRegressor(n_estimators=500,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    verbosity=0)
```

```python
regessors={
    "LinearRegression":lr,
    "KNeighborsRegressor":knn,
    "DecisionTreeRegressor":dt,
    "RandomForestRegressor":rf,
    "SVR":svc,
    "Ridge":rd,
    "Lasso":las,
    "GradientBoostingRegressor":gb,
    "XGBRegressor":xgb

}
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
def train_regressor(regressor, x_train, y_train, x_test, y_test, name):
  # Train the regessor
  regressor.fit(x_train, y_train)

  # make prediction
  y_pred=regressor.predict(x_test)

  # Evaluate the regressor
  mse=mean_squared_error(y_test,y_pred)
  mae=mean_absolute_error(y_test,y_pred)
  r2=r2_score(y_test,y_pred)

  return mse, mae, r2
```

```python
# Lists to store mse, mae, r2
mse_scores = []
mae_scores = []
r2_scores = []

# Loop through each regressor
for name, reg in regessors.items():
  current_mse, current_mae, current_r2 = train_regressor(
        reg, x_train_scaled, y_train, x_test_scaled, y_test, name
    )

  # Print results
  print(f'Name: {name}')
  print(f'MSE: {current_mse:.4f}')
  print(f'MAE: {current_mae:.4f}')
  print(f'R2: {current_r2:.4f}')
  print('----------------------------')

  # Store results
  mse_scores.append(current_mse)
  mae_scores.append(current_mae)
  r2_scores.append(current_r2)
```

```python
from sklearn.pipeline import make_pipeline
```

```python
pipe1=make_pipeline(lr)
pipe2=make_pipeline(knn)
pipe3=make_pipeline(dt)
pipe4=make_pipeline(rf)
pipe5=make_pipeline(svc)
pipe6=make_pipeline(rd)
pipe7=make_pipeline(las)
pipe8=make_pipeline(gb)
pipe9=make_pipeline(xgb)



```

```python
pipe1.fit(x_train_scaled,y_train)
pipe2.fit(x_train_scaled,y_train)
pipe3.fit(x_train_scaled,y_train)
pipe4.fit(x_train_scaled,y_train)
pipe5.fit(x_train_scaled,y_train)
pipe6.fit(x_train_scaled,y_train)
pipe7.fit(x_train_scaled,y_train)
pipe8.fit(x_train_scaled,y_train)
pipe9.fit(x_train_scaled,y_train)
```

```python

y_pred1=pipe1.predict(x_test_scaled)
y_pred2=pipe2.predict(x_test_scaled)
y_pred3=pipe3.predict(x_test_scaled)
y_pred4=pipe4.predict(x_test_scaled)
y_pred5=pipe5.predict(x_test_scaled)
y_pred6=pipe6.predict(x_test_scaled)
y_pred7=pipe7.predict(x_test_scaled)
y_pred8=pipe8.predict(x_test_scaled)
y_pred9=pipe9.predict(x_test_scaled)
```

```python

print(r2_score(y_test,y_pred1))
print(r2_score(y_test,y_pred2))
print(r2_score(y_test,y_pred3))
print(r2_score(y_test,y_pred4))
print(r2_score(y_test,y_pred5))
print(r2_score(y_test,y_pred6))
print(r2_score(y_test,y_pred7))
print(r2_score(y_test,y_pred8))
print(r2_score(y_test,y_pred9))
```
